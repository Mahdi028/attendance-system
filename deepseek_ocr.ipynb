### OCR

```python
from transformers import AutoModel, AutoTokenizer, BitsAndBytesConfig
import torch

model_name = 'deepseek-ai/DeepSeek-OCR'
```

```python
qc = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_use_double_quant=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.float
)
```

```python
tokenizer = AutoTokenizer.from_pretrained(model_name, trust_remote_code=True)
model = AutoModel.from_pretrained(
    model_name, trust_remote_code=True,
    use_safetensors=True, device_map="auto",
    quantization_config=qc, torch_dtype=torch.float
)
model = model.eval()
```

```python
prompt = "<image>\n<|grounding|>Convert the document to markdown."
# Prompts examples:
# document: <image>\n<|grounding|>Convert the document to markdown.
# other image: <image>\n<|grounding|>OCR this image.
# without layouts: <image>\nFree OCR.
# figures in document: <image>\nParse the figure.
# general: <image>\nDescribe this image in detail.
# rec: <image>\nLocate <|ref|>xxxx<|/ref|> in the image.
```

```python
image_file = "/content/Screenshot 2026-11-24 232527.png"
from IPython.display import Image
Image(image_file, width=640)
```

```python
with torch.no_grad():
    res = model.infer(
        tokenizer,
        prompt=prompt,
        image_file=image_file,
        output_path=".",
        base_size=1024,
        image_size=1024,
        crop_mode=False,
        save_results=True,
        test_compress=True,
        eval_mode=False
    )
```

```python
# سلول کامل و مستقل — برای runtime جدید (نیازی به سلول قبلی ندارد)
!pip -q install fastapi uvicorn pyngrok nest-asyncio pydantic python-multipart

import shutil, tempfile, os, time
import nest_asyncio, threading, uvicorn
from fastapi import FastAPI, UploadFile, Form
from fastapi.responses import JSONResponse

# ۱) پیدا کردن مدل و توکنایزر (سلول‌های لود مدل نوتبوک باید قبلاً اجرا شده باشند)
MODEL = TOK = None
for name, val in list(globals().items()):
    if name.startswith("_"):
        continue
    if TOK is None and "Tokenizer" in type(val).__name__:
        TOK = val
    if MODEL is None and hasattr(val, "infer"):
        MODEL = val
assert MODEL is not None, "مدل پیدا نشد — سلول‌های لود مدل نوتبوک را اول اجرا کنید"
assert TOK is not None, "توکنایزر پیدا نشد — سلول‌های لود مدل نوتبوک را اول اجرا کنید"
print(f"مدل: {MODEL.__class__.__name__} | توکنایزر: {TOK.__class__.__name__}")

# ۲) تعریف اپ و endpoint نهایی از صفر — هیچ endpoint قدیمی اینجا وجود ندارد
app = FastAPI()
IMG_EXTS = (".jpg", ".jpeg", ".png")

def _read_newest_result(out_dir):
    files = sorted(
        (os.path.join(out_dir, f) for f in os.listdir(out_dir)
         if not f.lower().endswith(IMG_EXTS)),
        key=os.path.getmtime,
    )
    if files:
        with open(files[-1], encoding="utf-8", errors="ignore") as fh:
            return fh.read()
    return None

@app.post("/ocr")
async def ocr_endpoint(file: UploadFile, prompt: str = Form("<image>\nFree OCR.")):
    tmp = tempfile.NamedTemporaryFile(suffix=os.path.splitext(file.filename)[1], delete=False)
    shutil.copyfileobj(file.file, tmp)
    tmp.close()
    out_dir = tempfile.mkdtemp(prefix="ocr_out_")
    res = None
    try:
        import io as _io, contextlib as _cl
        captured = _io.StringIO()
        with _cl.redirect_stdout(captured):
            res = MODEL.infer(TOK, prompt=prompt, image_file=tmp.name,
                              output_path=out_dir, base_size=1024,
                              image_size=640, crop_mode=True, save_results=True,
                              test_compress=True, eval_mode=True)
        # مسیر ۱: خود infer رشته داد
        if not isinstance(res, str) or not res.strip():
            res = None
            # مسیر ۲: مدل نتیجه را در فایل نوشت
            res = _read_newest_result(out_dir)
        # مسیر ۳: نتیجه فقط در کنسول چاپ شده
        if res is None or not str(res).strip():
            printed = captured.getvalue()
            lines = [ln for ln in printed.splitlines()
                     if ln.strip() and not ln.startswith("==")
                     and "torch.Size" not in ln and "NO PATCHES" not in ln]
            res = "\n".join(lines) if lines else ""
    finally:
        os.unlink(tmp.name)
        shutil.rmtree(out_dir, ignore_errors=True)
    return JSONResponse({"text": str(res) if res else ""})

# ۳) اجرای سرور
nest_asyncio.apply()
def run_server():
    uvicorn.run(app, host="0.0.0.0", port=8000, log_level="warning")
threading.Thread(target=run_server, daemon=True).start()
time.sleep(3)
print("Server ready on port 8000")

# ۴) تونل ngrok
from pyngrok import ngrok
ngrok.set_auth_token("3MegYrzZLrlhqH4jybkgVOGHVjT_63YyTou4uNtT7T5LXJADW")
url = ngrok.connect(8000)
print("=" * 50)
print("PUBLIC URL:", url.public_url)
print("=" * 50)
```
